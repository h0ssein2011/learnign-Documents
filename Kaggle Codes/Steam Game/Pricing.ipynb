```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('steam_games_dataset.csv')
df['total_reviews'] = df['positive'] + df['negative']
df_filtered = df[df['total_reviews'] > 0].copy()
df_filtered['sentiment_ratio'] = df_filtered['positive'] / df_filtered['total_reviews']
df_filtered['price_usd'] = df_filtered['initialprice'] / 100.0

bins = [-1, 0, 9.99, 19.99, 39.99, 59.99, np.inf]
labels = ['Free ($0)', '< $10', '$10-$19.99', '$20-$39.99', '$40-$59.99', '$60+']
df_filtered['price_tier'] = pd.cut(df_filtered['price_usd'], bins=bins, labels=labels)

plt.figure(figsize=(10, 6))
sns.boxplot(x='price_tier', y='sentiment_ratio', data=df_filtered, palette='viridis', showfliers=False)
plt.title('Review Sentiment Ratio by Base Price Tier', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Base Price Tier (USD)', fontsize=12)
plt.ylabel('Sentiment Ratio (Positive / Total)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('price_vs_sentiment.png')
```
